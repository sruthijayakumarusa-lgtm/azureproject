## DimUser

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

import os
import sys

project_path=os.path.join(os.getcwd(),'..','..')
sys.path.append(project_path)

from utils.transformations import reusable
```

```python
df_user= spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format","parquet")\
            .option("cloudFiles.schemaLocation","abfss://silver@rawadlsproject.dfs.core.windows.net/DimUser/checkpoint")\
            .option("schemaEvolutionMode","rescue")\
            .load("abfss://bronze@rawadlsproject.dfs.core.windows.net/DimUser")
```

```python
#transformations in dimuser
#1.All user names in uppercase
df_user=df_user.withColumn("user_name",upper(col("user_name")))

```

```python
df_user_obj=reusable()
df_user = df_user_obj.dropColumns(df_user,['_rescued_data'])
df_user=df_user.dropDuplicates(["user_id"])
```

```python
df_user.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation","abfss://silver@rawadlsproject.dfs.core.windows.net/DimUser/checkpoint")\
    .trigger(once=True)\
    .option("path","abfss://silver@rawadlsproject.dfs.core.windows.net/DimUser/data")\
    .toTable("spotify_cata.silver.DimUser")
```

## DimTrack

```python
df_track= spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format","parquet")\
            .option("cloudFiles.schemaLocation","abfss://silver@rawadlsproject.dfs.core.windows.net/DimTrack/checkpoint")\
            .option("schemaEvolutionMode","rescue")\
            .load("abfss://bronze@rawadlsproject.dfs.core.windows.net/DimTrack")
```

```python
df_track=df_track.withColumn("duration_flag",when(col('duration_sec')< 150,"low")\
                                .when(col('duration_sec')<300,"medium")\
                                .otherwise("high"))
```

```python
df_track=df_track.withColumn("track_name",regexp_replace(col('track_name'),'-',' '))
```

```python
df_track_obj=reusable()
df_track = df_track_obj.dropColumns(df_track,['_rescued_data'])
```

```python
df_track.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation","abfss://silver@rawadlsproject.dfs.core.windows.net/DimTrack/checkpoint")\
    .trigger(once=True)\
    .option("path","abfss://silver@rawadlsproject.dfs.core.windows.net/DimTrack/data")\
    .toTable("spotify_cata.silver.DimTrack")
```
